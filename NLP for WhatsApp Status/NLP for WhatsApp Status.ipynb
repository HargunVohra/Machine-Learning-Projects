```python
happy = happy.drop_duplicates(subset = "content", keep = "first")
sad = sad.drop_duplicates(subset = "content", keep = "first")
angry = angry.drop_duplicates(subset = "content", keep = "first")
```

```python
frames = [happy, sad, angry]
df = pd.concat(frames)
```

```python
df.info()
```

```python
df = df.drop_duplicates(subset = "content", keep = "first")
```

```python
df["sentiment"].value_counts()
```

```python
df.replace({"sentiment": {"happy": 1, "angry": 0, "sad": 2}}, inplace=True)

import pandas as pd
pd.set_option('future.no_silent_downcasting', True)
```

```python
df['sentiment'].value_counts()
```

```python
import re
def html(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, "", text)
df["content"] = df["content"].apply(html)
```

```python
def lower(text):
    return text.lower()
df["content"] = df["content"].apply(lower)
```

```python
def special(text):
        x=''
        for i in text:
            if i.isalnum():
                x=x+i
            else:
                x=x+' '
        return x
df['content']=df['content'].apply(special)
```

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
import nltk
from nltk.stem.porter import PorterStemmer
```

```python
def r_stopwords(text):
    x=[]
    for i in text.split():
        
        if i not in stopwords.words('english'):
            x.append(i)
    y=x[:]
    x.clear()
    return y
df['content']=df['content'].apply(r_stopwords)
```

```python
def join_back(list_input):
    return " ".join(list_input)
df['content']=df['content'].apply(join_back)
```

```python
ps= PorterStemmer()
y=[]
def stem_words(text):
    for i in text:
        y.append(ps.stem(i))
    z=y[:]
    y.clear()
    return z
df['content']=df['content'].apply(stem_words)
```

```python
def joinback2(list_input):
    return "".join(list_input)
df['content']=df['content'].apply(joinback2)
```

```python
df
```

```python
X = df["content"]
```

```python
y=df.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test = train_test_split(X,y,test_size=0.3) 
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=1500)
```

```python
X_train = cv.fit_transform(X_train).toarray()
X_test=cv.transform(X_test).toarray()
```

```python
from sklearn.naive_bayes import MultinomialNB
model=MultinomialNB()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred2=model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print("Multinomial",accuracy_score(y_test,y_pred2))
```
